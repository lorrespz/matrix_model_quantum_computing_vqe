In this notebook, we rerun some of the VQE experiments for bosonic SU(2) matrix model at $\Lambda = 4$ (using different variants of EvolvedOperatorAnsatz) at different Estimator seeds.

```python
#pip install qiskit = 1.3.0
#pip install qiskit_aer = 0.15
import sys
sys.path.append('../../utility')
from vqe_run import *
from qc_ansatze import *
from L4_evop_func import *
```

# VQE individual runs

```python
#storing values
counts = []
values = []
def store_intermediate_result(eval_count, parameters, mean, std):
    counts.append(eval_count)
    values.append(mean)
    
def run_qve_w_specified_optimizer(optimizer, ansatz, iters, seed):
    algorithm_globals.random_seed = seed
    #estimator
    noiseless_estimator = AerEstimator(
        run_options={"seed": seed, "shots": 1024},
        transpile_options={"seed_transpiler": seed})
    opt = optimizer(maxiter = iters)
    vqe = VQE(noiseless_estimator, ansatz, optimizer=opt, callback=store_intermediate_result)
    result = vqe.compute_minimum_eigenvalue(operator=H4q).eigenvalue.real
    print(f"VQE result: {result:.5f}")
    return result
```

```python
seeds_list_1 = [28, 170, 225, 188]
seeds_list_2 = np.arange(10, 15)
seeds = []
[seeds.append(seeds_list_1[i]) for i in range(len(seeds_list_1))]
[seeds.append(seeds_list_2[i]) for i in range(len(seeds_list_2))]
seeds
```

#  lambda = 0.2: ev_op_Hp15 (COBYLA)

```python
l4_l02_evop = L4_evop(0.2)
```

```python
H4q = l4_l02_evop.H4q
```

```python
l4_l02_evop.ansatz_list[0], l4_l02_evop.ansatz_names[0]
```

```python
r_cobyla=[]
seeds = [28, 170, 225, 188, 18, 38, 48, 58]
for i in range(len(seeds)):
    print(f'At step {i}, with {seeds[i]}')
    counts = []
    values = []
    t0 = time.time()
    result = run_qve_w_specified_optimizer(COBYLA, l4_l02_evop.ansatz_list[0], seeds[i], 650)
    t1 = time.time()
    print(f'Length of this optimization {len(values)}, time taken = {np.round(t1-t0,3)} \n')
    counts_a = counts
    values_a = values 
    r_cobyla.append(pd.DataFrame({f'seed_{seeds[i]}': values_a}))
    
```

```python
df1 = pd.concat([r_cobyla[i] for i in range(len(r_cobyla))], axis = 1)
df1.to_csv('results_seeds/l4_l02_op_ev_Hp15_cobyla_seeds.csv')
```
