# SU(2) model, $\Lambda = 4$: Result Preprocessing

Notebook created by HLD for the work arXiv: 2503.13368 [quant-ph, hep-th]

In this notebook, we process and select the best results for those  VQE experiments that were rerun (using different Estimator seeds) in another notebook. 

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings('ignore')
```

```python
E_02 = 3.13406
E_05 = 3.29894

E_10 = 3.52625
E_20 = 3.89548
```

```python
def choose_best_seed(df, Ee):
    lv = []
    if df.columns[0] == 'Unnamed: 0':
        df.drop(df.columns[0], axis = 1, inplace = True)
    for i in range(len(df.columns)):
        last_v = df[df.columns[i]].to_numpy()[-1]
        lv_e = last_v - Ee
        if lv_e >0:
            lv.append(lv_e)
        else:
            lv.append(2.0)
    b_v = np.argmin(lv)
    
    return df.columns[b_v], df[df.columns[b_v]].to_numpy()[-1]
```
