In this notebook, we process the results of several VQE experiment reruns and choose the best results. 

```python
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings('ignore')
```

```python
E_02 = 3.14808
E_05 = 3.36254
E_10 = 3.69722
E_20 = 4.26795
```

```python
def choose_best_seed(df, Ee):
    lv = []
    if df.columns[0] == 'Unnamed: 0':
        df.drop(df.columns[0], axis = 1, inplace = True)
    for i in range(len(df.columns)):
        last_v = df[df.columns[i]].dropna().to_numpy()[-1]
        lv_e = last_v - Ee
        if lv_e >0:
            lv.append(lv_e)
        else:
            lv.append(2.0)
    b_v = np.argmin(lv)
    print(lv)
    return df.columns[b_v], df[df.columns[b_v]].dropna().to_numpy()[-1]
```

# lambda = 0.2

```python
df_02_ct= pd.read_csv('results/qve_l2_l0.2_twolocal_cobyla.csv')
df_c0a = pd.read_csv('results_seeds/l2_l02_tl_Ry_c_cobyla_seeds.csv')
df_c0d = pd.read_csv('results_seeds/l2_l02_tl_RyY_c_cobyla_seeds.csv')
```

```python
choose_best_seed(df_c0a, E_02)
```

```python
choose_best_seed(df_c0d, E_02)
```

```python
df_02_ct['ansatz_0a'] = df_c0a['seed_48']
df_02_ct['ansatz_0d'] = df_c0d['seed_48']
```

```python
df_02_ct.drop(['Unnamed: 0'], axis = 1, inplace = True)
df_02_ct.to_csv('results/qve_l2_l0.2_twolocal_cobyla.csv')
```
