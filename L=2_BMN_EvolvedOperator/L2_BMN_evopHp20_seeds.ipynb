# Supersymmetric SU2 matrix model: EvolvedOperatorAnsatz rerun

Notebook created by HLD for the work arXiv: 2503.13368 [quant-ph, hep-th]

The best ansatz `ev-op-Hp20' is rerun with different hardward configs to get the best results. 

```python
import sys
import pandas as pd
sys.path.append('../../utility')
import time
from L2_evop_func import *
```

```python
from qiskit_algorithms import VQE
from qiskit_algorithms.optimizers import SPSA
from qiskit_algorithms.utils import algorithm_globals 
from qiskit_aer.primitives import Estimator as AerEstimator
import warnings
warnings.filterwarnings("ignore")

def run_qve_w_specified_optimizer(operator, optimizer, ansatz, seed = 170, iterations = 250):
    #storing values
    counts = []
    values = []
    def store_intermediate_result(eval_count, parameters, mean, std):
        counts.append(eval_count)
        values.append(mean)
        
    algorithm_globals.random_seed = seed
    noiseless_estimator = AerEstimator(
        run_options={"seed": seed, "shots": 1024},
        transpile_options={"seed_transpiler": seed},)
    opt = optimizer(maxiter = iterations)
    vqe = VQE(noiseless_estimator, ansatz, optimizer=opt, callback=store_intermediate_result)
    result = vqe.compute_minimum_eigenvalue(operator).eigenvalue.real
    print(f"VQE result: {result:.6f}")
    return result, values

def vqe_rerun(seeds, H, optimizer, ansatz, iterations = 2000):
    r_res=[]
    for i in range(len(seeds)):
        print(f'At step {i}, with seed {seeds[i]}')
        t0 = time.time()
        result, values = run_qve_w_specified_optimizer(H, optimizer, ansatz, seeds[i], iterations)
        t1 = time.time()
        print(f'Length of this optimization {len(values)}, time taken = {np.round(t1-t0,3)} \n')
        r_res.append(pd.DataFrame({f'seed_{seeds[i]}': values}))
    return r_res
```

# lambda = 0.2

```python
l2_02_eo = L2_supersymmetric_EvOp('02')
```

```python
seeds = [10, 20, 30, 40, 50, 155, 125]
H4q = l2_02_eo.H4q
ansatz = l2_02_eo.ansatz_list[1]
print(f'{l2_02_eo.ansatz_names[1]}')
r_res = vqe_rerun(seeds, H4q, SPSA, ansatz, iterations = 2000)

df = pd.concat([r_res[i] for i in range(len(r_res))], axis = 1)
df.to_csv('results_seeds/l2_l02_evopHp20_spsa_seeds_3.csv')
```
