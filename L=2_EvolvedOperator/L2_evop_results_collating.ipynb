```python
import pandas as pd
import numpy as np
import time

import sys
sys.path.append('../../utility')
from L4_evop_data_collate import *

E_02 = 3.14808
E_05 = 3.36254
E_10 = 3.69722
E_20 = 4.26795
```

```python
def load_and_split(link):
    df = pd.read_csv(link)
    df.drop(['Unnamed: 0'], axis = 1, inplace=True)
    df_c = df[df.columns[:9]]
    df_s = df[df.columns[9:]]
    col_names = ['ev_op_r', 'ev_op_r3', 'ev_op_H', 'ev_op_H_2f',
       'ev_op_H_3f', 'ev_op_Hp', 'ev_op_Hp2', 'ev_op_Hp3',
       'ev_op_Hp4']
    df_c.columns = col_names
    df_s.columns = col_names
    return df_c, df_s
```
